```python
def fun_bmi(person, height, weight):
    # 输入姓名，身高（m），体重（kg）
    bmi = weight / (height * height)
    print(f"{person}的bmi指数为: {bmi:.2f}") # 保留两位小数

    if bmi < 18.5:
        print('体重过轻')
    elif 18.5 <= bmi < 24.9:  
        print('正常范围')
    elif 24.9 <= bmi < 29.9: 
        print('超重') 
    else:              
        print('肥胖')
```
