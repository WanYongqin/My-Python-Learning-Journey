```python
# 创建文件夹及子文件夹存放数据
import os

# 定义基础路径，方便以后修改
base_dir = r"E:\study\Python\L10"

for i in range(1, 121):
    for k in range(1, 11):
        # 拼接完整的嵌套路径
        folderpath = os.path.join(base_dir, f"Sub{i:03d}", f"Run{k:02d}")
        
        # 若目录已存在不会报错
        os.makedirs(folderpath, exist_ok=True)

print("文件夹创建完毕！")
```
